```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
import ast
from pathlib import Path

# Configuración visual
# plt.style.use('dark_background')
sns.set_context("notebook")
```

```python
def load_spectra(sample):
    intensity = json.loads(sample['Intensity'].values[0])
    velocity = json.loads(sample['Velocity (mm/s)'].values[0])
    return velocity, intensity
```

```python
ruta_csv = Path('Data/data.csv')
df = pd.read_csv(ruta_csv)
df.sample(1)
```

```python
fig, ax1 = plt.subplots(figsize=(12, 8))

# Distribución de Clase Mineralógica

clases = df['folder'].value_counts().head(20)
sns.barplot(x=clases.values, y=clases.index, ax=ax1)
ax1.set_title("Distribución de clases")
ax1.set_xlabel("Número de Espectros en el dataset")
ax1.set_ylabel("Clase")
plt.show()
```

```python
piroxene = df[df['folder'] == 'OH'].sample(20)
velocity, intensity = load_spectra(piroxene)
plt.figure(figsize=(10, 6))
plt.plot(velocity, intensity, linestyle='-', color='blue')
plt.title("Espectro de un Piroxeno")
plt.xlabel("Velocidad (mm/s)")
plt.ylabel("Intensidad")    
```
